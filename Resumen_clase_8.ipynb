Entrenando modelos: Regresión lineal

La regresión lineal es un modelo que permite hacer predicciones encontrando una relación entre una variable y otra. Por ejemplo, podemos utilizarla para predecir la satisfacción de vida de una población a partir de su PIB.

La ecuación normal es una solución matemática que permite calcular directamente los parámetros del modelo que minimizan el error entre las predicciones y los valores reales.La regresión polinomial permite adaptar la regresión lineal cuando la relación entre los datos no sigue una línea recta, utilizando diferentes grados de polinomios para representar mejor los datos.Regresion lineal..jpg
