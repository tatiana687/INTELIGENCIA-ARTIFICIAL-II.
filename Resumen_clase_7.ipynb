Objetivo de negocio y tubería de datos

Todo proyecto de Inteligencia Artificial debe comenzar con un objetivo de negocio claro, no directamente con una línea de código. En este caso, el objetivo es construir un modelo que pueda predecir el precio medio de las viviendas.

La predicción del modelo será utilizada posteriormente por un sistema de análisis de inversiones, que ayudará a tomar decisiones sobre dónde destinar el capital. Por eso, es importante definir desde el principio cuál es el objetivo del proyecto y cómo se medirá su éxito.

Esto nos lleva al concepto de tubería de datos (pipeline). De forma sencilla, podemos imaginarla como una línea de ensamblaje: los datos pasan por diferentes etapas, y en cada una se procesan o transforman hasta obtener un resultado final útil.“Lo que entendí es que un proyecto de inteligencia artificial primero debe tener un objetivo claro. En este caso, queremos predecir el precio de las viviendas para apoyar decisiones de inversión. Para lograrlo usamos una tubería de datos, que es como una línea de ensamblaje donde los datos pasan por diferentes etapas hasta obtener un resultado.
