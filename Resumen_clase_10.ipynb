Árboles de decisión – Algoritmo CART

Los árboles de decisión son modelos que toman decisiones dividiendo los datos en diferentes partes mediante reglas. El algoritmo CART permite construir árboles tanto para clasificación como para regresión.

Árbol de regresión: se utiliza cuando queremos predecir un valor numérico, por ejemplo, el precio de una vivienda.

Regresión sin restricciones: si dejamos que el árbol crezca demasiado, puede aprender demasiado bien los datos de entrenamiento, incluyendo sus errores y detalles específicos. Esto puede producir alta varianza y hacer que el modelo funcione peor con datos nuevos.

Modelo regularizado: limita la complejidad del árbol, por ejemplo, controlando su profundidad o realizando poda. Esto ayuda a evitar el sobreajuste y a que el modelo generalice mejor.

¿Cómo superar la alta varianza?

Una forma de reducir la alta varianza es limitar la complejidad del árbol mediante regularización, por ejemplo, reduciendo su profundidad o podándolo. En regresión, el árbol predice valores numéricos. Si dejamos que crezca demasiado puede aparecer alta varianza, porque aprende demasiado los datos de entrenamiento. Para solucionarlo podemos regularizar o limitar el árbol y así mejorar su capacidad para trabajar con datos nuevos.CART.jpg
