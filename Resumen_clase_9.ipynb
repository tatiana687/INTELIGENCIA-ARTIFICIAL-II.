Máquinas de Vectores de Soporte (SVM)

Las SVM son modelos de Machine Learning que buscan separar diferentes clases de datos. Su idea principal es encontrar una frontera de decisión que deje la mayor distancia posible entre los grupos. A esta distancia se le llama margen máximo.

La imagen presenta cuatro módulos:

1. Clasificación lineal
Busca una línea o frontera que separe las diferentes clases. La idea es encontrar “la calle más ancha”, es decir, la separación con el mayor margen posible.

2. Clasificación no lineal
Cuando los datos no pueden separarse con una línea recta, se utiliza el kernel, que permite trabajar con relaciones más complejas entre los datos.

3. Regresión
Las SVM también pueden utilizarse para realizar predicciones de valores, no solamente para clasificar grupos.

4. Under the Hood
Hace referencia a comprender qué ocurre matemáticamente dentro del modelo, es decir, cómo funcionan sus cálculos y fundamentos.Lo que entendí es que las SVM buscan separar diferentes grupos de datos tratando de dejar el mayor margen posible entre ellos. Cuando los datos se pueden separar con una línea se usa clasificación lineal, y cuando son más complejos se pueden utilizar kernels. Además, las SVM también pueden utilizarse para regresión.”
