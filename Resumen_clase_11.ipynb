1. Los de diversidad aleatoria (Paralelos):*
Son trucos para que los árboles no sean todos iguales.
Parches aleatorios:* Toman un poquito de datos Y un poquito de características al azar.
Subespacios aleatorios:* Toman TODOS los datos, pero eligen características al azar.
Extra Trees (Árboles Extra):* Es el más extremo y rápido. No busca el mejor punto para cortar, corta al azar. Por eso es rapidísimo.

*2. Los de corrección secuencial (Boosting) Aquí ya no es en paralelo, es uno detrás de otro, aprendiendo del error.
Es como un equipo que se pasa el examen corregido.

Paso 1: Entrenan un modelo base.
Paso 2: El modelo se equivoca en algunos datos. Entonces CASTIGAN esos datos, les suben el peso.
Paso 3: El siguiente modelo está OBLIGADO a fijarse en esos casos difíciles que fallaron.
 Y así se repite. Cada vez la línea que divide los datos se curva más y se adapta mejor.



Hay dos hermanos con estrategias diferentes:

Hermano 1 - AdaBoost: Es el obsesivo de los pesos. Su estrategia es: "a cada dato que fallé, le subo el peso numérico".
Hermano 2 - Gradient Boosting: Es el cirujano. No toca pesos. Su estrategia es: "voy a entrenar al siguiente modelo para que adivine EXACTAMENTE el error que dejó el anterior". Solo corrige el residuo.

*3. El truco para que sea ultra rápido (Escalamiento):*
Para poder con terabytes de datos, usan el *truco de los contenedores o histogramas* (LightGBM, XGBoost).
En vez de usar números con decimales infinitos, meten todo en cajitas con números enteros. Pierden un poquito de precisión, pero ganan 100 veces más velocidad.


> "Para terminar, vimos tres familias. Primero, los que crean diversidad al azar como Parches y Extra Trees que cortan al azar para ser más rápidos. Segundo, entramos al Boosting que es secuencial: un modelo aprende del error del anterior. AdaBoost lo hace subiendo el peso de los datos fallados y Gradient Boosting corrigiendo el residuo exacto. Y por último, para escalarlo, se usa la técnica de histogramas que empaqueta los datos en contenedores para ser cientos de veces más rápido."
