Fases de evaluación

1. Entrenamiento: se entrena el modelo utilizando el conjunto principal de datos, para que aprenda los patrones.

2. Validación: se utilizan datos diferentes para ajustar los hiperparámetros y mejorar la configuración del modelo.

3. Prueba final: se evalúa el desempeño del modelo utilizando un conjunto de datos separado, que no se había utilizado durante el entrenamiento.

¿Qué riesgos tienen los modelos no supervisados?

Los modelos no supervisados trabajan con datos que no tienen etiquetas conocidas. Por eso, uno de los principales riesgos es que el modelo encuentre patrones o grupos que realmente no sean significativos. También puede ser difícil comprobar si los resultados obtenidos son correctos, ya que no existe una respuesta previamente definida para comparar.Primero se entrena el modelo, después se valida para ajustar sus parámetros y finalmente se prueba con datos que no había visto.
