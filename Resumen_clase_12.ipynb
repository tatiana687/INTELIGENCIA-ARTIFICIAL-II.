Primero, qué es el PCA y cómo funciona
Lo que entendí es que el PCA sirve para hacer los datos más chiquitos sin perder lo importante. Tiene 3 pasos: primero Centrar, que es mover todos los puntos al centro para quitar el desorden. Segundo, Aplicar SVD, que es buscar hacia dónde se estiran más los datos, esas son las direcciones más importantes. Y tercero, Proyectar, que es aplastar los datos viejos sobre esas direcciones nuevas y más pequeñas.


Segundo, cómo elijo qué PCA usar Matriz de decisiones

No todos los PCA sirven para lo mismo, por eso hay una matriz de decisiones:Si tengo poca memoria en mi PC, uso el Incremental, porque trabaja por mini-lotes, de a poquitos.Si necesito rapidez, uso el Aleatorio, que es estocástico, o sea que adivina rápido y bien sin calcular todo perfecto.Si tengo una dimensión masiva, con miles de datos, uso el de Proyección, porque es el más eficiente.

El PCA a veces no sirve cuando los datos están enrollados como un caracol en forma de G. Para eso están los otros 3, y en la imagen los comparan con el mismo dato:MDS: Intenta mantener las distancias reales, pero aplasta el rollo y mezcla los colores.Isomap: Es más inteligente, mide la distancia caminando por encima del rollo, como una hormiguita, y por eso lo logra desenrollar en una barra ordenada.t-SNE: No le importa la forma, le importa agrupar. Rompe el rollo en pedacitos para que los vecinos sigan juntos y se vean los grupos de colores.
